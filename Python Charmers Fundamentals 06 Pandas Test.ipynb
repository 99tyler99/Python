```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US = df[(df['origin_country']=='US') & (df['dest_country']!='US')]
flights_outof_US = flights_outof_US[['passengers','unique_carrier_name','origin_country','dest_country']]
flights_outof_US
```

```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US_grouped = flights_outof_US.groupby('unique_carrier_name').sum('passengers')
flights_outof_US_grouped = flights_outof_US_grouped.sort_values('passengers',ascending=False).head(10)
flights_outof_US_grouped
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
flights_outof_US_with_dest = flights_outof_US.groupby(['unique_carrier_name', 'dest_country']).agg({'dest_country':'count'})
flights_outof_US_with_dest = flights_outof_US_with_dest.rename(columns={'dest_country':'dest_country_count'})
flights_outof_US_with_dest = flights_outof_US_with_dest.reset_index()
flights_outof_US_with_dest
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
joined_flights = pd.merge(flights_outof_US_grouped,flights_outof_US_with_dest,on='unique_carrier_name',how='left')
joined_flights = joined_flights.sort_values(by=['unique_carrier_name','dest_country_count'], ascending=[True,False])
joined_flights
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
most_popular_destinations = joined_flights.groupby('unique_carrier_name').max()
most_popular_destinations = most_popular_destinations.reset_index()
populardropped = most_popular_destinations[['unique_carrier_name','dest_country_count']]
populardropped
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
finaljoin = pd.merge(joined_flights,populardropped,on=['unique_carrier_name','dest_country_count'],how='inner')
finaljoin = finaljoin[['unique_carrier_name','dest_country','dest_country_count']]
finaljoin
```

```python
# 15: For the top 15 international countries, find the percentage of class F passengers traveling to/from the top 10 US cities for international travel 
# (e.g. if only cities A, B, and C flew into Aruba, the sum of the Aruba row/column would be 1). Plot the result as a heatmap.
question15_pivoted = pd.pivot_table(question14_joined,values='passengers_y',index=['US_city','intl_country'],aggfunc='sum',dropna=True)
question15_pivoted
question15_nozero = question15_pivoted.loc[(question15_pivoted!=0).any(axis=1)]
question15_nozero
```

```python
# 15: For the top 15 international countries, find the percentage of class F passengers traveling to/from the top 10 US cities for international travel 
# (e.g. if only cities A, B, and C flew into Aruba, the sum of the Aruba row/column would be 1). Plot the result as a heatmap.
question15_citylist = pd.pivot_table(question14_joined,values='passengers_y',index='US_city',aggfunc='sum',dropna=True)
question15_citylist = question15_citylist.sort_values('passengers_y',ascending=False).head(10)
question15_citylist = question15_citylist.reset_index()
US_citylist = question15_citylist['US_city'].to_list()
US_citylist
```

```python
# 15: For the top 15 international countries, find the percentage of class F passengers traveling to/from the top 10 US cities for international travel 
# (e.g. if only cities A, B, and C flew into Aruba, the sum of the Aruba row/column would be 1). Plot the result as a heatmap.
question15_countrylist = pd.pivot_table(question14_joined,values='passengers_y',index='intl_country',aggfunc='sum',dropna=True)
question15_countrylist = question15_countrylist.sort_values('passengers_y',ascending=False).head(15)
question15_countrylist = question15_countrylist.reset_index()
countrylist = question15_countrylist['intl_country'].tolist()
countrylist
```

```python
# 15: For the top 15 international countries, find the percentage of class F passengers traveling to/from the top 10 US cities for international travel 
# (e.g. if only cities A, B, and C flew into Aruba, the sum of the Aruba row/column would be 1). Plot the result as a heatmap.
question15_reset = question15_reset.reset_index()
question15_crosstab = pd.crosstab(question15_reset['intl_country'],question15_reset['US_city'],question15_reset['passengers_y'],aggfunc='sum')
question15_crosstab
```

```python
# 15: For the top 15 international countries, find the percentage of class F passengers traveling to/from the top 10 US cities for international travel 
# (e.g. if only cities A, B, and C flew into Aruba, the sum of the Aruba row/column would be 1). Plot the result as a heatmap.
def to_percent_of_row(question15_crosstab):
    row_sums = question15_crosstab.sum(axis=1)
    return question15_crosstab.div(row_sums, axis=0)

question15_df = to_percent_of_row(question15_crosstab)
question15_df
```
