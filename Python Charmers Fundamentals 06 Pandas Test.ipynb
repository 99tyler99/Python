```python
# 5. Find the top 5 carriers for median route distance (i.e., for all origin-destination pairs that each carrier has, find the median distance after removing duplicates).
df[['distance','unique_carrier_name','origin','dest']]
```

```python
# 5. Find the top 5 carriers for median route distance (i.e., for all origin-destination pairs that each carrier has, find the median distance after removing duplicates).
subset_cols = ['unique_carrier_name','origin','dest']
df_unique = df.drop_duplicates(subset=subset_cols)

df_unique = df_unique[['unique_carrier_name','distance','origin','dest']]
df_unique
```

```python
# 5. Find the top 5 carriers for median route distance (i.e., for all origin-destination pairs that each carrier has, find the median distance after removing duplicates).
grouped_median = df_unique.groupby('unique_carrier_name').median('distance') 
grouped_median = grouped_median.rename(columns={'distance':'median_distance'})
grouped_median.sort_values('median_distance',ascending=False).head()
```

```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
df_cargo = df[['freight','mail','distance','unique_carrier_name']]
df_cargo
```

```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
df_cargo['cargo'] = df_cargo['freight'] + df_cargo['mail']
df_cargo
```

```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
grouped_df_cargo = df_cargo.groupby('unique_carrier_name').agg({'cargo': 'sum', 'distance': 'mean'})
grouped_df_cargo = grouped_df_cargo.rename(columns={'cargo': 'total_cargo','distance': 'mean_distance'})
grouped_df_cargo.sort_values('total_cargo',ascending=False).head(10)
```

```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US = df[(df['origin_country']=='US') & (df['dest_country']!='US')]
flights_outof_US = flights_outof_US[['passengers','unique_carrier_name','origin_country','dest_country']]
flights_outof_US
```

```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US_grouped = flights_outof_US.groupby('unique_carrier_name').sum('passengers')
flights_outof_US_grouped = flights_outof_US_grouped.sort_values('passengers',ascending=False).head(10)
flights_outof_US_grouped
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
flights_outof_US_with_dest = flights_outof_US.groupby(['unique_carrier_name', 'dest_country']).agg({'dest_country':'count'})
flights_outof_US_with_dest = flights_outof_US_with_dest.rename(columns={'dest_country':'dest_country_count'})
flights_outof_US_with_dest
```
