```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
df_cargo = df[['freight','mail','distance','unique_carrier_name']]
df_cargo
```

```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
df_cargo['cargo'] = df_cargo['freight'] + df_cargo['mail']
df_cargo
```

```python
# 6. Find the total cargo transported (mail + freight) and mean distance traveled for the 10 carriers that transported the most cargo.
grouped_df_cargo = df_cargo.groupby('unique_carrier_name').agg({'cargo': 'sum', 'distance': 'mean'})
grouped_df_cargo = grouped_df_cargo.rename(columns={'cargo': 'total_cargo','distance': 'mean_distance'})
grouped_df_cargo.sort_values('total_cargo',ascending=False).head(10)
```

```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US = df[(df['origin_country']=='US') & (df['dest_country']!='US')]
flights_outof_US = flights_outof_US[['passengers','unique_carrier_name','origin_country','dest_country']]
flights_outof_US
```

```python
# 7. Which 10 carriers flew the most passengers out of the United States to another country?
flights_outof_US_grouped = flights_outof_US.groupby('unique_carrier_name').sum('passengers')
flights_outof_US_grouped = flights_outof_US_grouped.sort_values('passengers',ascending=False).head(10)
flights_outof_US_grouped
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
flights_outof_US_with_dest = flights_outof_US.groupby(['unique_carrier_name', 'dest_country']).agg({'dest_country':'count'})
flights_outof_US_with_dest = flights_outof_US_with_dest.rename(columns={'dest_country':'dest_country_count'})
flights_outof_US_with_dest = flights_outof_US_with_dest.reset_index()
flights_outof_US_with_dest
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
joined_flights = pd.merge(flights_outof_US_grouped,flights_outof_US_with_dest,on='unique_carrier_name',how='left')
joined_flights = joined_flights.sort_values(by=['unique_carrier_name','dest_country_count'], ascending=[True,False])
joined_flights
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
most_popular_destinations = joined_flights.groupby('unique_carrier_name').max()
most_popular_destinations = most_popular_destinations.reset_index()
populardropped = most_popular_destinations[['unique_carrier_name','dest_country_count']]
populardropped
```

```python
# 8. For each of the carriers found in #7, find the most popular destination country outside of the United States.
finaljoin = pd.merge(joined_flights,populardropped,on=['unique_carrier_name','dest_country_count'],how='inner')
finaljoin = finaljoin[['unique_carrier_name','dest_country','dest_country_count']]
finaljoin
```

```python
# 9. For each of the carriers found in #7, find the total number of passengers flown on international flights to/from the destinations in #8 or the United States. 
# Note that this dataset only has data for flights with an origin and/or destination of the United States.
destinations_passengers = flights_outof_US.groupby(['unique_carrier_name','dest_country']).sum('passengers')
destinations_passengers
```

```python
# 9. For each of the carriers found in #7, find the total number of passengers flown on international flights to/from the destinations in #8 or the United States. 
# Note that this dataset only has data for flights with an origin and/or destination of the United States.
question9_outofUS = pd.merge(finaljoin,destinations_passengers,on=['unique_carrier_name','dest_country'],how='inner')
question9_outofUS = question9_outofUS.rename(columns={'passengers':'passengers_outofUS','dest_country': 'join_country'})
question9_outofUS = question9_outofUS[['unique_carrier_name','join_country','passengers_outofUS']]
question9_outofUS
```

```python
# 9. For each of the carriers found in #7, find the total number of passengers flown on international flights to/from the destinations in #8 or the United States. 
# Note that this dataset only has data for flights with an origin and/or destination of the United States.
flights_to_US = df[(df['origin_country']!='US') & (df['dest_country']=='US')]
flights_to_US = flights_to_US[['passengers','unique_carrier_name','origin_country','dest_country']]
flights_to_US = flights_to_US.rename(columns={'origin_country':'join_country'})
flights_to_US
```
